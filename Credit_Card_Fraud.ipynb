# HANDLE CLASS IMBALANCE

# 30. Separate Features and Target

```python
# Separate independent features from the targeet variable
x = df.drop("Class", axis=1)

# Store the target variable
y = df["Class"]

# Display the shape of features and target
print("Features shape:", x.shape)
print("Target shape:", y.shape)
```

# 31. Split the Dataset into Training and Testing Sets

```python
# Import train_test_split for splitting the dataset
from sklearn.model_selection import train_test_split

# Split the dataset into training and testing sets
# stratify=y keeps the same class proportion in both sets
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.2, random_state=42, stratify=y)

# Display the shapes of training and testing sets
print("x_train shape:", x_train.shape)
print("x_test shape:", x_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

# 31. Check Class Distribution Before SMOTE

```python
# Check the class distribution in the training data
print("Class distribution before SMOTE:")
print(y_train.value_counts())
```

# 32. Import SMOTE

```python
# Import SMOTE for handling class imbalance
from imblearn.over_sampling import SMOTE

# Print message confirmation
print("Library is imported")
```

# 33. Apply SMOTE Only on Training Data

```python
# Create the SMOTE object 
smote = SMOTE(random_state=42)

# Apply SMOTE only to the training data
x_train_resampled, y_train_resampled= smote.fit_resample(x_train, y_train)

# Display the shapes after SMOTE
print("x_train after SMOTE:", x_train_resampled.shape)
print("y_train after SMOTE:", y_train_resampled.shape)
```

# 34. Check Class Distribution After SMOTE

```python
# Check the class distribution after applying SMOTE
print("Class distribution after SMOTE:")
print(y_train_resampled.value_counts())
```

# 35. Visualize Balanced Training Data

```python
# Visualize the class distribution after SMOTE
plt.figure(figsize=(6,4))

sns.countplot(x=y_train_resampled)

# Add title and labels 
plt.title("Class Distribution After SMOTE")
plt.xlabel("Transaction Class")
plt.ylabel("Number of Transaction")

# Display the chart
plt.show()
```

# 36. Import StandardScaler

```python
# Create a StandardScaler object
scaler = StandardScaler()
```

# 38. Scale Training Data

```python
# Fit the scaler on the training data and transform it
x_train_scaled = scaler.fit_transform(x_train_resampled)
```

# MODEL BUILDING

# 42. Import Logistic Regression

```python
# Import Logistic Regression for classification
from sklearn.linear_model import LogisticRegression

# Print message confirmation
print("Library is imported")
```

# 43. Create the Model

```python
# Create the Logistic Regression model
model = LogisticRegression(random_state=42, max_iter=1000)
```

# 44. Train the Model

```python
# Train the model using the balanced and scaled training data
model.fit(x_train_scaled, y_train_resampled)
```

# 45. Make Predictions

```python
# Predict the classes for the test dataset
y_pred = model.predict(x_test_scaled)

# Display the first 10 predictions
y_pred[:10]
```

# 46. Predict Probabilities

```python
# Predict the probability of each transaction belonging to each class
y_pred_proba = model.predict_proba(x_test_scaled)

# Display the first 5 probability predictions
y_pred_proba[:5]
```

# ROC Curve & AUC Score

# 57. Import ROC and AUC Metrics

```python
# Import ROC curve and AUC score
from sklearn.metrics import roc_curve, roc_auc_score

# Print the message confirmation
print("Libray is imported")
```

# 58. Calculate ROC Curve

```python
# Calculate the probability of the positive class (Fraud)
y_prob = model.predict_proba(x_test_scaled)[:,1]

# Calculate False Positive Rate and True Positive Rate
fpr, tpr, thresholds = roc_curve(y_test, y_prob)
```

# 59. Calculate AUC Score

```python
# Calculate the Area Under the ROC Curve
auc_score = roc_auc_score(y_test, y_prob)

# Display the AUC score
print("ROC-AUC Score:", auc_score)
```

# 60. Plot ROC Curve

```python
# Create a figure for the ROC curve 
plt.figure(figsize=(8,5))

# plot the ROC curve
plt.plot(fpr, tpr, label=f"LogictisRegression (AUC = {auc_score:.4f})")

# Plot the random classifier line
plt.plot([0,1],[0,1], linestyle="--")

# Add title and labels
plt.title("ROC Curve")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")

# Add legend
plt.legend()

# Display the chart
plt.show()
```

# FEATURE IMPORTANCE

# 61. Extract Feature Coefficients

```python
# Create a DataFrame containing features names and theior coefficients
feature_importance = pd.DataFrame({"Feature": x.columns, "Coefficient": model.coef_[0]})

# Display the feature coefficients
feature_importance.head()
```

# 62. Calculate Absolute Feature Importance

```python
# Calculate the absolute value of coefficients
feature_importance["Importance"] = feature_importance["Coefficient"].abs()

# Sort features by importance
feature_impotance = feature_importance.sort_values(by="Importance", ascending=False)

# Display the most important features
feature_importance.head(10)
```

# 63. Visualize Top 10 Important Features

```python
# Select the top 10 most important features
top_features = feature_importance.head(10)

# Create a bar plot
plt.figure(figsize=(10,6))

sns.barplot(x="Importance", y="Feature", data=top_features)

# Add title and labels
plt.title("Top 10 Important Features")
plt.xlabel("Feature Importance")
plt.ylabel("Features")

# Display the chart
plt.show()
```

# 64. Display Positive and Negative Coefficients

```python
# Display the 10 features with the highest positive coefficients
positive_features = feature_importance.sort_values(by="Coefficient", ascending= False).head(10)

print("Top Positive Features:")
print(positive_features[["Feature","Coefficient"]])
```

```python
# Display  the 10 features with the lowest coefficients
negative_features = feature_importance.sort_values(by="Coefficient", ascending= True).head(10)

print("Top Negative Features:")
print(negative_features[["Feature","Coefficient"]])
```

# 65. Project Conclusion

 The Credit Card Fraud Detection project was developed to identify fraudulent and legitimate credit card transactions using machine learning.


 The dataset was first explored and cleaned by checking missing values, duplicate records, data types, and statistical information. Exploratory Data Analysis was then performed to understand transaction patterns and the distribution of normal and fraudulent transactions.

    
 Since fraudulent transactions are significantly fewer than normal transactions, SMOTE (Synthetic Minority Over-sampling Technique) was applied only to the training data to handle class imbalance and prevent data leakage. The features were then standardized using StandardScaler.

    
 A Logistic Regression classification model was trained and evaluated using Accuracy, Precision, Recall, F1-Score, Confusion Matrix, and ROC-AUC Score.

    
 Feature importance analysis was also performed using Logistic Regression coefficients to identify the features that had a stronger influence on fraud prediction.
    

 Overall, this project demonstrates a complete machine learning workflow for a real-world fraud detection problem, including data understanding, data cleaning, EDA, class imbalance handling, feature scaling, model building, model evaluation, and feature importance analysis
